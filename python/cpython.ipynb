# Введение в Python и CPython

## Вопросы

### 1. Что такое Python?
На Python пишут программы, сайты и инструменты для работы с данными.

### 2. Сколько существует реализаций Python? Какая самая популярная?
Реализаций несколько: CPython, PyPy, Jython, IronPython, MicroPython и другие. Самая распространённая из них CPython.

### 3. На каком языке написан CPython?
В основном на языке C.

### 4. Кто создал CPython?
Гвидо ван Россум, создатель языка Python.

### 5. Почему Python считается быстрым?
Python не всегда быстрый. Но многие библиотеки выполняют тяжёлые вычисления на C и C++, поэтому их работа может быть быстрой.

### 6. Где находится интерпретатор CPython на моём компьютере?
У меня установлен CPython 3.12.0. Путь к файлу:

`C:\Users\АТ\AppData\Local\Programs\Python\Python312\python.exe`


### 7. Что находится в папке Include?
В ней лежат заголовочные файлы C с расширением `.h`. В основной папке я нашёл 74 файла, а также две папки: `cpython` и `internal`.

### 8. Где найти исходный код CPython?
На GitHub: https://github.com/python/cpython. В файле `README.rst` есть сведения о сборке, тестировании и документации.

### 9. Как CPython выполняет программу?
Сначала переводит код в байт-код. Затем виртуальная машина CPython выполняет эти команды.

### 10. Как запустить файл через CPython?
`python example.py`

### 11. Можно ли запускать текстовый файл `.txt` через Python?
Да, если внутри написан Python-код. Я проверил это на файле `example.txt`.

### 12. Как указать путь к интерпретатору и файлу?
В PowerShell:

```powershell
& "C:\Users\АТ\AppData\Local\Programs\Python\Python312\python.exe" "C:\Users\АТ\example.py"
```


### 13. Чем PyPy отличается от CPython?
PyPy использует JIT-компиляцию, чтобы ускорять часто выполняемый код. CPython 3.12 обычно исполняет байт-код через интерпретатор.

### 14. Почему PyPy подходит не для всех проектов?
Некоторые библиотеки с ним несовместимы. В коротких программах он также не всегда быстрее.

### 15. Где скачать PyPy?
На сайте https://www.pypy.org/download.html.

### 16. Как установить PyPy после скачивания?
Я скачал архив для Windows 64-bit и распаковал его на диск `C:`. После этого запустил `pypy.exe`.

### 17. Как запустить файл через PyPy?
В PowerShell:

```powershell
& "C:\pypy3.12-v8.0.1-win64\pypy3.12-v8.0.1-win64\pypy.exe" ".\example_pypy.py"
```

### 18. Почему PyPy иногда быстрее CPython?
JIT превращает часто повторяющиеся части программы в машинный код. Процессор выполняет его напрямую.


## Практические задания

### 1. Проверка CPython
Проверил через PowerShell. Установлен CPython 3.12.0.

```powershell
python --version
python -c "import platform; print(platform.python_implementation())"
```

Получил `Python 3.12.0` и `CPython`.

### 2. Структура CPython
Открыл папку `Include`. В ней 74 файла `.h` без учёта вложенных папок. Нашёл `Python.h` и прочитал `README.rst` на GitHub.

### 3. Запуск через CPython
Создал `example.txt` с таким кодом:

```python
print("Hello from CPython!")
```

Запустил его через `python example.txt`, затем переименовал в `example.py` и повторил запуск. В обоих случаях получил `Hello from CPython!`.


### 4. Запуск через PyPy
Распаковал PyPy 8.0.1 для Windows. Создал `example_pypy.py`:

```python
print("Hello from pypy!")
```

Запустил файл через `pypy.exe`. Получил `Hello from pypy!`.

### 5. Сравнение CPython и PyPy
Для проверки запустил один и тот же код в двух интерпретаторах:

```python
import time

start_time = time.time()
total = 0
for i in range(1, 10000000):
    total += i
end_time = time.time()

print("Result:", total)
print("Execution time:", end_time - start_time, "seconds")
```

| Интерпретатор | Время выполнения |
|---|---:|
| CPython 3.12.0 | 1,583 с |
| PyPy 8.0.1 | 0,090 с |

Оба раза сумма получилась `49999995000000`. В этом запуске PyPy был примерно в 17,7 раза быстрее. При повторном тесте время может отличаться.
